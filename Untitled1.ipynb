## Rename 'datetime' to 'ds' and 'close' to 'y' to prepare the DataFrame for Prophet.

```python
df = df.rename(columns={'datetime': 'ds', 'close': 'y'})
```

## Create a new DataFrame `df3` with only the 'ds' and 'y' columns required by Prophet.

```python
df3 = df[['ds','y']]
```

## Initialize and fit the Prophet model using the prepared DataFrame `df3`.

```python
m = Prophet()
m.fit(df3)
```

## Create a future DataFrame for forecasting, extending 365 periods beyond the last date.

```python
future = m.make_future_dataframe(periods=365)
```

## Display the generated `future` DataFrame.

```python
future
```

## Generate predictions using the fitted Prophet model and display the tail of the forecast.

```python
forecast = m.predict(future)
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail()
```

## Plot the forecast results, showing the actuals, predictions, and uncertainty intervals.

```python
fig1 = m.plot(forecast)
```

## Plot the individual components of the Prophet forecast (trend, weekly, daily seasonality).

```python
fig2 = m.plot_components(forecast)
```

## Generate an interactive Plotly chart of the forecast.

```python
plot_plotly(m, forecast)
```
